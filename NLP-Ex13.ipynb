```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'boy' | 'girl'
V -> 'likes'
""")

parser = nltk.ChartParser(grammar)

sentence = "the boy likes the girl".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
