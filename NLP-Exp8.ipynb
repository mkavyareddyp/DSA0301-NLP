```python
from collections import defaultdict, Counter

training = [
    [('I', 'PRP'), ('like', 'VBP'), ('dogs', 'NNS')],
    [('I', 'PRP'), ('like', 'VBP'), ('cats', 'NNS')],
    [('She', 'PRP'), ('likes', 'VBZ'), ('dogs', 'NNS')]
]

tag_counts = defaultdict(Counter)

for sentence in training:
    for word, tag in sentence:
        tag_counts[word][tag] += 1

sentence_to_tag = ["I", "like", "dogs"]

print("Predicted tags:")
for word in sentence_to_tag:
    if word in tag_counts:
        tag = tag_counts[word].most_common(1)[0][0]
        print(word, "->", tag)
    else:
        print(word, "->", "UNKNOWN")
```
