```python
singular_subjects = ["boy", "girl", "cat", "dog"]
plural_subjects = ["boys", "girls", "cats", "dogs"]

singular_verbs = ["runs", "eats", "plays", "walks"]
plural_verbs = ["run", "eat", "play", "walk"]


def check_agreement(sentence):
    words = sentence.lower().split()

    if len(words) != 3:
        return False

    article = words[0]
    subject = words[1]
    verb = words[2]

    if subject in singular_subjects and verb in singular_verbs:
        return True

    if subject in plural_subjects and verb in plural_verbs:
        return True

    return False


sentence = "The boy runs"

if check_agreement(sentence):
    print("Accepted: Subject and verb agree.")
else:
    print("Rejected: Subject and verb do not agree.")
```
