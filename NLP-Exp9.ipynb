```python
import re

def tag(word):
    if re.match(r'.*ing$', word):
        return "VBG"
    elif re.match(r'.*ly$', word):
        return "RB"
    elif re.match(r'.*s$', word):
        return "NNS"
    elif word[0].isupper():
        return "NNP"
    else:
        return "NN"

words = ["Kavya", "playing", "quickly", "books"]

for word in words:
    print(word, "->", tag(word))
```
