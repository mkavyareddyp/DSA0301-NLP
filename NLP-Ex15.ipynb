```python
import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6]
NP -> 'John' [0.4]
VP -> V NP [1.0]
Det -> 'the' [1.0]
N -> 'dog' [1.0]
V -> 'sees' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "John sees the dog".split()

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```
