```python
def plural(noun):
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    else:
        return noun + "s"


words = ["cat", "box", "bus", "baby", "book"]

for word in words:
    print(word, "->", plural(word))
```

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat' | 'mouse'
V -> 'chased'
""")

parser = nltk.ChartParser(grammar)

sentence = "the cat chased the mouse".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
