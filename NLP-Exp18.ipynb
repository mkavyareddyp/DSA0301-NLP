```python
import re

def parse_fopc(expression):
    expression = expression.strip()

    pattern = r'^(forall|exists)\s+([a-zA-Z]+)\s*\((.+)\)$'
    match = re.match(pattern, expression)

    if not match:
        return "Invalid FOPC expression"

    quantifier = match.group(1)
    variable = match.group(2)
    predicate = match.group(3)

    return {
        "Quantifier": quantifier,
        "Variable": variable,
        "Predicate": predicate
    }


expression = "forall x (Human(x) -> Mortal(x))"

result = parse_fopc(expression)

print("Expression:", expression)
print("Parsed Expression:")

if isinstance(result, dict):
    for key, value in result.items():
        print(key + ":", value)
else:
    print(result)
```
