```python
def dialog_act(sentence):
    s = sentence.lower()

    if "?" in sentence:
        return "QUESTION"
    elif s.startswith(("hi", "hello", "hey")):
        return "GREETING"
    elif s.startswith(("thank", "thanks")):
        return "THANKING"
    elif s.startswith(("bye", "goodbye")):
        return "GOODBYE"
    else:
        return "STATEMENT"


dialog = [
    "Hello!",
    "How are you?",
    "I am fine.",
    "Thank you!",
    "Goodbye!"
]

for sentence in dialog:
    print(sentence, "->", dialog_act(sentence))
```
