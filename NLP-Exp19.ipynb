```python
from nltk.corpus import wordnet as wn
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

def lesk(word, sentence):
    word = word.lower()
    sentence_words = set(sentence.lower().split())

    best_sense = None
    max_overlap = 0

    for sense in wn.synsets(word):
        definition_words = set(sense.definition().lower().split())
        overlap = len(sentence_words.intersection(definition_words))

        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = sense

    return best_sense


sentence = "He sat on the bank of the river"
word = "bank"

sense = lesk(word, sentence)

print("Sentence:", sentence)
print("Word:", word)

if sense:
    print("Meaning:", sense.definition())
    print("Synset:", sense.name())
else:
    print("Meaning not found")
```
